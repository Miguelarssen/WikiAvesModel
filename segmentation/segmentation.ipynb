```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display
from pathlib import Path

import bird_segmentation as bs

NPY_FILE = Path(r"c:\Users\Miguel\Desktop\EstudosUTFPR\WikiAvesModel\dataset\bacurau-ocelado\bacurau-ocelado_333.npy")
SR = 44110

S = np.load(NPY_FILE).astype(np.float32)
if S.ndim == 3: S = S.squeeze()

imagem_original, imagem_limpa, binary, mask, segs, S_seg = bs.segment_spectrogram(S)

audio_orig = bs.spec_to_audio(S, sr=SR)
audio_seg  = bs.spec_to_audio(S_seg, sr=SR)

plt.figure(figsize=(15, 5))
plt.subplot(131); plt.title("Original"); plt.imshow(S, aspect='auto', origin='lower', cmap='magma')
plt.subplot(132); plt.title("Máscara"); plt.imshow(mask, aspect='auto', origin='lower', cmap='gray')
plt.subplot(133); plt.title("Segmentado (Pixel-wise)"); plt.imshow(S_seg, aspect='auto', origin='lower', cmap='magma')
plt.show()

display(Audio(audio_orig, rate=SR))
display(Audio(audio_seg, rate=SR))

```
